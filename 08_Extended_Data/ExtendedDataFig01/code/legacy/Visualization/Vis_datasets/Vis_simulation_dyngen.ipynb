```python
import pandas as pd
import numpy as np
import math
import matplotlib.pyplot as pl
import matplotlib.pyplot as plt
import os
import scanpy as sc
import scvelo as scv
from scvelo.core import sum
import celldancer.cdplt as cdplt
```

```python
para = pd.read_csv('/media/liyaru/LYR/Benchmark/dyngen/Parameters_dyngen.csv')
para.columns
```

## pca

```python
nrow = 7
ncol = 6
fig, axs = plt.subplots(nrow, ncol, figsize=(3 * ncol, 3 * nrow))
f = 0

for i in para.columns[1:]:
    print(i)
    adata = sc.read_h5ad('/media/liyaru/LYR/Benchmark/Github/data_dyngen/anndata/' + i + "/anndata.h5ad")
    # print(adata)

    r = int(f/ncol)
    c = f % ncol
    ax = axs[r, c]
    
    fig_i = sc.pl.pca(adata,ax=ax, show=False,
                      legend_loc=False,frameon=False,
                      colorbar_loc=None,
                      color = 'sim_time')
    
    ax.axis('off')
    ax.set_title(i,fontdict={'size': 10}) 
    ax.tick_params(
                top="off",
                bottom="off",
                left="off",
                right="off",
                labelleft="off",
                labelbottom="off",
            )

    f = f+1
    

plt.savefig('/media/liyaru/LYR/Benchmark/Fig/10.dyngen_simu_pca.pdf',bbox_inches='tight')
plt.savefig('/media/liyaru/LYR/Benchmark/Fig/9.dyngen_simu_pca.png',bbox_inches='tight',format='png',dpi=300)

plt.show()
```

## velocity stream

```python
nrow = 7
ncol = 6
fig, axs = plt.subplots(nrow, ncol, figsize=(5 * ncol, 5 * nrow))
f = 0

for i in para.columns[1:]:
    print(i)
    adata = sc.read_h5ad('/media/liyaru/LYR/Benchmark/Github/data_dyngen/anndata/' + i + "/anndata.h5ad")
    # print(adata)

    r = int(f/ncol)
    c = f % ncol
    ax = axs[r, c]
    adata.layers['ground_truth_velocity'] = adata.layers['ground_truth_velocity'].A

    scv.tl.velocity_graph(adata,vkey='ground_truth_velocity',basis='pca',n_jobs=10)
    
    fig_i = scv.pl.velocity_embedding_stream(adata,ax=ax, show=False,
                      # legend_loc='none',
                      frameon=False,
                      color = 'sim_time',
                      vkey='ground_truth_velocity',basis='pca')
    
    ax.axis('off')
    ax.set_title(i) 
    ax.tick_params(
                top="off",
                bottom="off",
                left="off",
                right="off",
                labelleft="off",
                labelbottom="off",
            )

    f = f+1
    

plt.savefig('/media/liyaru/LYR/Benchmark/Fig/10.dyngen_simu_pca_stream.svg',bbox_inches='tight',format='svg')
plt.savefig('/media/liyaru/LYR/Benchmark/Fig/10.dyngen.cd_simu_pca_stream.png',bbox_inches='tight',format='png',dpi=200)

plt.show()
```

## gene scatter

```python
import celldancer.utilities as cdutil

fig = plt.figure(figsize=(30,100))
f = 1

for i in para.columns[1:]:
    print(i)
    adata = sc.read_h5ad('/media/liyaru/LYR/Benchmark/Github/data_dyngen/anndata/' + i + "/anndata.h5ad")
    # print(adata)

    cdutil.adata_to_df_with_embed(adata,
                                us_para=['Mu', 'Ms'],
                                cell_type_para='sim_time',
                                embed_para='X_pca',
                                save_path='/media/liyaru/LYR/Benchmark/dyngen/df/' + i + '.csv'
                                )


    df = pd.read_csv('/media/liyaru/LYR/Benchmark/dyngen/df/' + i + '.csv')
    gene = df['gene_name'].unique()[0]

    ax = fig.add_subplot(42, 6, f)
    cdplt.scatter_gene(
            ax=ax,
            x='splice',
            y='unsplice',
            colors='clusters',
            cellDancer_df = df,
            custom_xlim=None,
            custom_ylim=None,
            alpha=0.8, 
            s = 10,
            velocity=False,
            gene=gene)
    
    ax.axis('off')
    ax.set_title(i,fontdict={'size': 10}) 

    f = f+1

plt.savefig('/media/liyaru/LYR/Benchmark/Fig/10.dyngen_simu_gene.pdf',bbox_inches='tight')
plt.savefig('/media/liyaru/LYR/Benchmark/Fig/10.dyngen_simu_gene.png',bbox_inches='tight',format='png',dpi=300)
plt.show()
```
